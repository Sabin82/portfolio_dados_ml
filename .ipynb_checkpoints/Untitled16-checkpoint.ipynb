```python
import pandas as pd

# Lista de gastos mensais dos clientes do banco
gastos = [120, 150, 200, 110, 180, 10000]

# 1. Transformando a lista em uma Series do Pandas
df_gastos = pd.Series(gastos)
print(df_gastos.mean())
print(df_gastos.median())
print(df_gastos.describe())
```

```python
import matplotlib.pyplot as plt

# Cria o gráfico de caixa (Boxplot)
plt.boxplot(df_gastos)
plt.title("Detecção Visual de Outliers de Faturamento")
plt.ylabel("Valores em R$")
plt.show()

```
