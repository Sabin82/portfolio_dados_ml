```python
import sqlite3
import pandas as pd
import numpy as np

# 1. Cria a conexão com o novo banco SQLite na memória
conexao = sqlite3.connect(':memory:')
ponteiro = conexao.cursor()

# 2. Cria a tabela de transações do banco digital
ponteiro.execute('''
    CREATE TABLE transacoes (
        id_transacao INTEGER PRIMARY KEY,
        cliente TEXT,
        valor REAL,
        data_envio TEXT
    )
''')

# 3. Gerando o histórico estável do Bruno (Média ~50, Desvio Padrão ~15)
# Usando o numpy para simular dados reais baseados na engenharia mecânica/estatística
np.random.seed(42)
historico_valores = np.random.normal(loc=50, scale=15, size=100).round(2)
dados_historico = [(i, 'Bruno', float(val), '2026-09-19') for i, val in enumerate(historico_valores)]
ponteiro.executemany('INSERT INTO transacoes VALUES (?, ?, ?, ?)', dados_historico)

# 4. Inserindo as 3 novas compras de HOJE para o teste de fraude
compras_hoje = [
    (101, 'Bruno', 80.0, '2026-09-21'),   # Compra A
    (102, 'Bruno', 100.0, '2026-09-21'),  # Compra B
    (103, 'Bruno', 20.0, '2026-09-21')    # Compra C
]
ponteiro.executemany('INSERT INTO transacoes VALUES (?, ?, ?, ?)', compras_hoje)
conexao.commit()

print("🚀 Novo ambiente criado com sucesso! Pronto para a segunda-feira de Estatística.")

```
