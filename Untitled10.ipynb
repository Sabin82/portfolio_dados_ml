```python
def media_sem_none(valores):
    # Remove valores None
    valores_limpos = [v for v in valores if v is not None]
    
    # Calcula a média
    media = sum(valores_limpos) / len(valores_limpos)
    print(valores_limpos)
    
    return media

 
# Exemplo de uso
#dados = [10, 20, None, 40, 50, None]
#print(media_sem_none(dados))
```

```python
dados = [10, 20, None, 40, 50, None]
print(media_sem_none(dados))
```

```python
def contar_frequencia(valores):
    frequencia = {}
    for v in valores:
        if v in frequencia:
            frequencia[v] += 1
        else:
            frequencia[v] = 1
    return frequencia
 
# Exemplo de uso
dados = ["maçã", "banana", "maçã", "laranja", "banana", "maçã"]
print(contar_frequencia(dados))
```
