```python
import pandas as pd

# Dicionário com todos os dados preenchidos corretamente
dados_comportamento = {
    'tempo_app': [30, 45, 60, 90, 120],
    'valor_gasto': [100, 150, 220, 310, 430],
    'idade': [25, 40, 31, 22, 35]
}

# Transformando em um DataFrame do Pandas
df_banco = pd.DataFrame(dados_comportamento)

# Calculando a matriz de correlação de Pearson
matriz_correlacao = df_banco.corr()
print(matriz_correlacao)
```

```python
import sqlite3

# Cria o banco e insere os dados brutos com nulos e ruídos
conexao = sqlite3.connect(':memory:')
ponteiro = conexao.cursor()
ponteiro.execute('''
    CREATE TABLE logs_aplicativo (
        id_log INTEGER PRIMARY KEY,
        usuario TEXT,
        tempo_sessao_minutos REAL
    )
''')
dados_logs = [
    (1, 'Carlos', 15.5), (2, 'Ana', None), (3, 'Bruno', 22.0),
    (4, 'Diana', -999.0), (5, 'Eduardo', 18.2), (6, 'Fabiana', 12.0)
]
ponteiro.executemany('INSERT INTO logs_aplicativo VALUES (?, ?, ?)', dados_logs)
conexao.commit()

```

```python
import pandas as pd
query = """
SELECT*
FROM logs_aplicativo
"""
tabela = pd.read_sql_query(query, conexao)
tabela
```

```python
import numpy as np
tabela2 = tabela.replace(-999, np.nan)
df = pd.DataFrame(tabela2)
colseparada = df['tempo_sessao_minutos']
colseparada.median()
```

```python
df.dropna().to_csv('dataset_treino_ia.csv', index=False)
```
