```python
import pandas as pd
import numpy as np # Dica: você pode usar np.nan para substituir o -999
import math 

leituras = [12.5, -999, 14.2, None, 15.8, -999]

# 1. Transforme a lista em uma Series do Pandas
serie_sensores = pd.Series(leituras)
serie_sensores2 = [a for a in serie_sensores if not math.isnan(a)]
for b in serie_sensores2[:]:
    if b == -999:
        serie_sensores2.remove(b)
print (pd.Series(serie_sensores2).mean())

# 2. Como você vai substituir os valores -999 por nulo e calcular a média?
# Dica: o Pandas tem a função .replace(-999, np.nan)
```

```python
import sqlite3
import pandas as pd

# 1. Cria a conexão com o banco SQLite na memória RAM
conexao = sqlite3.connect(':memory:')
ponteiro = conexao.cursor()

# 2. Cria a tabela de sensores da fábrica
ponteiro.execute('''
    CREATE TABLE sensores (
        id_sensor INTEGER PRIMARY KEY,
        temperatura REAL,
        status TEXT
    )
''')

# 3. Insere dados de teste (Sensores acima e abaixo de 85 graus)
dados_sensores = [
    (101, 72.3, 'Ativo'),    # Normal
    (102, 89.5, 'Ativo'),    # CRÍTICO! (Maior que 85 e Ativo)
    (103, 91.2, 'Inativo'),  # Quente, mas está desligado/inativo
    (104, 45.0, 'Ativo'),    # Frio
    (105, 95.8, 'Ativo'),    # CRÍTICO! (Maior que 85 e Ativo)
    (106, 86.0, 'Ativo')     # CRÍTICO! (Maior que 85 e Ativo)
]
ponteiro.executemany('INSERT INTO sensores VALUES (?, ?, ?)', dados_sensores)
conexao.commit()

print("Bancada de testes de SQL pronta!")
```

```python
import sqlite3
import pandas as pd

# 1. Cria a conexão com o banco SQLite na memória RAM
conexao = sqlite3.connect(':memory:')
ponteiro = conexao.cursor()
# Criando as tabelas para o Exercício 4 (Produtos e Defeitos)
ponteiro.execute('''
    CREATE TABLE produtos (
        id_produto INTEGER PRIMARY KEY,
        nome_produto TEXT
    )
''')

ponteiro.execute('''
    CREATE TABLE defeitos (
        id_defeito INTEGER PRIMARY KEY,
        id_produto INTEGER,
        descricao_defeito TEXT
    )
''')

# Inserindo dados de teste
ponteiro.executemany('INSERT INTO produtos VALUES (?, ?)', [
    (1, 'Eixo Motor'),
    (2, 'Engrenagem Cônica'),
    (3, 'Placa de Circuito'),
    (4, 'Sensor Ultrassônico')
])

ponteiro.executemany('INSERT INTO defeitos VALUES (?, ?, ?)', [
    (10, 1, 'Fissura na base por estresse térmico'),
    (11, 3, 'Curto-circuito na trilha de alimentação'),
    (12, 1, 'Desvio dimensional acima de 0.05mm')
])
conexao.commit()
print("Tabelas 'produtos' e 'defeitos' criadas com sucesso!")
```

```python
query = """
SELECT produtos.nome_produto, defeitos.descricao_defeito
FROM produtos
LEFT JOIN defeitos ON produtos.id_produto = defeitos.id_produto;
"""
table = pd.read_sql_query(query, conexao)
table.to_csv('relatorio_defeitos_fabrica.csv', index=False) 
print(table)
```
