```python
import re
from urllib.request import urlopen

def read_url(url):
    """Função oficial do Data 8 (UC Berkeley) para ler textos da web"""
    return re.sub('\\s+', ' ', urlopen(url).read().decode())
```

```python
def find_numbers (start, stop):
    result = []
    for i in range (start, stop + 1):
            if ((i % 7)==0 and (i % 5)!=0):
                result.append(str(i))
    return (','.join(result))

print (find_numbers (100, 200))      
```

```python
def dic_potencia (fim):
    a = dict()
    for i in range (1, fim + 1):
        a[i] = i ** 2
    return a
print (dic_potencia (100))
        
```
