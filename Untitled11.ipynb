```python
import sqlite3
import pandas as pd

# 1. Cria a conexão com um banco de dados SQLite na memória RAM
conexao = sqlite3.connect(':memory:')
ponteiro = conexao.cursor()

# 2. Cria a tabela de funcionários
ponteiro.execute('''
    CREATE TABLE funcionarios (
        id INTEGER PRIMARY KEY,
        nome TEXT,
        salario REAL
    )
''')

# 3. Insere os dados de teste (com salários acima/abaixo de 4000 e nulos)
dados_funcionarios = [
    (1, 'Ana', 3500.0),
    (2, 'Bruno', 4200.0),
    (3, 'Carlos', None),
    (4, 'Diana', 5000.0),
    (5, 'Eduardo', 6000.0),
    (6, 'Fabiana', None)
]
ponteiro.executemany('INSERT INTO funcionarios VALUES (?, ?, ?)', dados_funcionarios)
conexao.commit()

print("Banco de dados pronto para o treino!")
```
