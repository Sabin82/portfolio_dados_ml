```python
import sqlite3
import pandas as pd

# 1. Cria uma conexão com um banco de dados na memória do PC
conn = sqlite3.connect(':memory:')

# 2. Criando os dados fictícios (Agora com as listas numéricas corretas!)
dados_usuarios = {
    'id_usuario': [1, 2, 3],
    'nome': ['Ana', 'Bruno', 'Carina'],
    'cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte'],
    'data_cadastro': ['2026-01-10', '2026-01-12', '2026-01-15']
}

dados_pedidos = {
    'id_pedido': [101, 102, 103, 104],
    'id_usuario': [1, 1, 2, 99],
    'valor': [150.00, 80.00, 300.00, 45.00],
    'data_pedido': ['2026-02-01', '2026-02-03', '2026-02-04', '2026-02-05']
}

# 3. Salvando os dados como tabelas SQL dentro do banco em memória
pd.DataFrame(dados_usuarios).to_sql('usuarios', conn, index=False, if_exists='replace')
pd.DataFrame(dados_pedidos).to_sql('pedidos', conn, index=False, if_exists='replace')

print("Banco de dados SQL pronto para uso!")
```

```python
query = """
SELECT 
    u.nome, p.id_pedido, p.valor
FROM usuarios u
LEFT JOIN pedidos p 
ON u.id_usuario = p.id_usuario;
"""
```

```python
pd.read_sql_query(query, conn)
```

```python
query_agrupada = """
SELECT 
    u.nome, 
    SUM(p.valor) AS total_gasto
FROM usuarios u
LEFT JOIN pedidos p ON u.id_usuario = p.id_usuario
GROUP BY u.nome;
"""

pd.read_sql_query(query_agrupada, conn)
```
